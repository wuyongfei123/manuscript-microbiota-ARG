```python
# start coding here

import pandas as pd
from pathlib import Path
from qiime2 import Artifact
import biom
```

```python
div_path = Path(snakemake.input[0])

alpha_div_list =[]
for p in div_path.glob('*vector.qza'):
    tmp = Artifact.load(p)
    alpha_div_list.append(tmp.view(pd.Series))
pd.DataFrame(alpha_div_list).T.to_csv(snakemake.output['aplha_diversity_table'],sep='\t')
```

```python
from zipfile import ZipFile
for beta in ['weighted_unifrac_distance_matrix','jaccard_distance_matrix','bray_curtis_distance_matrix','unweighted_unifrac_distance_matrix']:
    p=div_path/f'{beta}.qza'
    tmp = ZipFile(p)
    for f in tmp.filelist:
        if 'data/distance-matrix.tsv' in f.filename:
            with open(snakemake.output[beta],'wb') as fo:
                fo.write(tmp.open(f,'r').read())
```
